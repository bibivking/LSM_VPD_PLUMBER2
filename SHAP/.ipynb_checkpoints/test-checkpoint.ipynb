```python
import pandas as pd
import numpy as np
import shap
import matplotlib.pyplot as plt
from sklearn.model_selection import GridSearchCV, train_test_split, KFold
from xgboost import XGBRegressor
import logging
import multiprocessing
import time
```

```python
num_cpus = multiprocessing.cpu_count()
print(f"Number of CPUs available: {num_cpus}")
```

### XGBoost

```python
def train_model(X: np.ndarray, y: np.ndarray, n_splits=5, test_size=0.2):
    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=42)
    
    # Set up XGBoost regressor
    regressor = XGBRegressor(objective="reg:squarederror", tree_method="gpu_hist",verbosity=2)

    # Define the grid search parameters
    param_grid = {
        'learning_rate': [0.1, 0.25, 0.35, 0.5],
        'max_depth': [7, 8, 9, 10, 11, 12],
        'subsample': [0.6, 0.8, 1], #0.2, 0.4, 
        'n_estimators': [100, 150, 200, 250, 300, 350, 400]
    }

    # Perform grid search with cross-validation
    grid_search = GridSearchCV(
        regressor, 
        param_grid, 
        cv=KFold(n_splits=n_splits, shuffle=True, random_state=42),
        scoring="neg_mean_squared_error",
        n_jobs=-1,    
        verbose=3,  # Show progress updates
        error_score="raise",
    )

    
    print(f"Starting GridSearchCV with {multiprocessing.cpu_count()} CPUs...")
    start_time = time.time()
    
    # Train the model using grid search
    grid_search.fit(X_train, y_train)

    end_time = time.time()
    print(f"GridSearchCV completed in {end_time - start_time:.2f} seconds.")


    # Get the best parameters from the grid search
    best_params = grid_search.best_params_
    print("Best hyperparameters found:", best_params)
    
    # Train the best model on the full training set
    best_regressor = grid_search.best_estimator_

    # Evaluate the final model on the test set
    final_predictions = best_regressor.predict(X_test)
    print("Out-of-sample predictions:", final_predictions)

    return best_regressor, X_train, X_test, y_train, y_test
```

```python
# logging.basicConfig(level=logging.DEBUG)

# Load dataset
var_input = pd.read_csv(
    '/srv/ccrc/LandAP/z5218916/script/PLUMBER2/LSM_VPD_PLUMBER2/txt/process1_output/Qle_all_sites.csv',
    usecols=['time', 'site_name', 'model_CABLE', 'VPD', 'obs_Tair', 'obs_Precip', 'obs_SWdown'],
    na_values=['']
)
var_input['CABLE_SMtop1m'] = pd.read_csv(
    '/srv/ccrc/LandAP/z5218916/script/PLUMBER2/LSM_VPD_PLUMBER2/txt/process1_output/SMtop1m_all_sites.csv',
    usecols=['CABLE_SMtop1m'], na_values=['']
)
var_input['obs_LAI'] = pd.read_csv(
    '/srv/ccrc/LandAP/z5218916/script/PLUMBER2/LSM_VPD_PLUMBER2/txt/process1_output/LAI_all_sites_parallel.csv',
    usecols=['obs_LAI'], na_values=['']
)

# Select a site for a trial
site_name = 'AU-Tum'
# var_select = var_input[var_input['site_name'] == site_name].dropna()
var_select = var_input

# Extract features and target
feature_names = ['VPD', 'obs_Tair', 'obs_Precip', 'obs_SWdown', 'CABLE_SMtop1m', 'obs_LAI']
X = var_select[feature_names].values
y = var_select['model_CABLE'].values

# Train the model and get results
model, X_train, X_test, y_train, y_test = train_model(X, y)
```

### SHAP

```python
# --- SHAP Analysis ---
# Use SHAP on the whole dataset (X)
explainer = shap.Explainer(model)
shap_values = explainer(X)  # Compute SHAP values for the entire dataset

# SHAP Summary Plot for the Whole Dataset
shap.summary_plot(shap_values, X, feature_names=feature_names)

# SHAP Dependence Plot (e.g., VPD vs Prediction)
shap.dependence_plot("VPD", shap_values, X, feature_names=feature_names)

# SHAP Force Plot for the First Prediction
shap.initjs()
shap.force_plot(explainer.expected_value, shap_values[0], feature_names=feature_names)
```
